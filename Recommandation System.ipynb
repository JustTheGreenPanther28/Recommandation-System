```python
# it's list under String
# to convert the string list into actual list; we will use library
import ast
```

```python
def convert(obj):
    category = []
    for i in ast.literal_eval(obj):
        category.append(i['name'])
    return category
```

```python
convert(movies.iloc[0].genres)
```

```python
movies['genres']=movies['genres'].apply(convert)
```

```python
movies = movies.rename(columns={'genres':'category'})
```

```python
movies.sample()
```

```python
movies['keywords'] = movies['keywords'].apply(convert)
```

```python
movies.head()
```

```python
# for cast (heros worked in company) we will choose top 3 of them
movies['cast'].iloc[0]
```

```python
def convert2(obj):
    name = []
    counter = 0
    for i in ast.literal_eval(obj):
        if counter<3:
            name.append(i['name'])
            counter+=1
        else :
            break
    return name
```

```python
movies['cast'] = movies['cast'].apply(convert2)
```

```python
movies.head()
```

```python
movies['crew'].iloc[0]
```

```python
def convert3(obj):
    directors = []
    for i in ast.literal_eval(obj):
        if i['job'] == 'Director':
            directors.append(i["name"])
    return directors
```

```python
movies['crew'] = movies['crew'].apply(convert3)
```

```python
# WE want full combine as one keyword to avoid ambiguity
# If there is name Ron Shelton and another Ron Arteta the model will get confused
# Ron Shelton ==> RonShelton
```

```python
def removeSpace(column):
    movies[column]=movies[column].apply(lambda x : [i.replace(" ","") for i in x])
```

```python
new_df = movies[['id','title','tags']]
```

```python
new_df
```

```python
# Convert tags into string
new_df['tags'] = new_df['tags'].apply(lambda x : " ".join(x))
```

```python
new_df['tags'][0]
```

```python
new_df['tags'] = new_df['tags'].apply(lambda x : x.lower())
```

```python
new_df.sample()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# it has problem ===> there are similar kind of words 
# activity and activities
# to remove this issue we use steming
# !pip install nltk
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
    y=[]
    
    for i in text.split():
        y.append(ps.stem(i))
        
    return " ".join(y)
```

```python
new_df['tags'] = new_df['tags'].apply(stem)
```

```python
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags']).toarray()
```

```python
vectors
```

```python
# we can actually what are these most frequent words
```

```python
cv.get_feature_names()
```

```python
cv.get_feature_names()[4999]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
np.where(similarity[5]>0.5)
# return index of movie whoes similarity score is > 0.5
```

```python
sorted(similarity[0]) # ascending order

sorted_val = sorted(similarity[0],reverse=True) # all data
sorted_val[:10]

# but indexes are lost so we use enumerate
```

```python
sorted(list(enumerate(similarity[0])),reverse = True) # sorted accroding to the index
sorted(list(enumerate(similarity[0])),reverse = True, key=lambda x:x[1]) # srrted according to second index
sorted(list(enumerate(similarity[0])),reverse = True, key=lambda x:x[1])[:5] # top five movies
```

```python
new_df
```

```python
# five similar movie
def recommend(movie_name):
    movie_index = new_df[new_df['title']==movie_name].index[0]
    distances = similarity[movie_index]
    top_5_recommended_movie_list = sorted(list(enumerate(distances)),reverse = True, key=lambda x:x[1])[1:6] # top five movies
    
    # printing recommended movies
    for i in top_5_recommended_movie_list : 
        print(new_df.iloc[i[0]].title)
```

```python
    recommend('Batman')
```
